```python
#assignment 6 \
import time 

def time_it(func):
    def wrapper (*args,**kwargs):
        start_time =time.time()
        result=func(*args,**kwargs)
        end_time =time.time()
        print (f"execution time :{end_time -start_time}")
        return result
    return wrapper

@time_it
def factoriaL (n):
    if n==0 :
        return 1
    else:
        return n * factoriaL(n-1)

print(factoriaL(10))
```

```python
#assignemnt 7
def repeat ( n):
    def decorator (func):
        def wrapper (*args, **kwargs):
            for _ in range (n):
                func(*args, **kwargs)
        return wrapper
    return decorator


@repeat(4)
def msg (str):
    print(str)

msg("amaan")        
```

```python
#asignment 8
def uppercase (func):
    def wrapper(*args,**kwargs):
        result=func(*args,**kwargs)
        return result.upper()
    return wrapper

def exclamation (func):
    def wrapper(*args,**kwargs):
        result=func(*args,**kwargs)
        return result + "!"
    return wrapper

@uppercase
@exclamation
def print_str (text):
    return text


print_str("amaan")
        
```
