```python
#assignmentt 7 1.

def insert_many_val(employees):
    conn = sqlite3.connect('test.db')
    c=conn.cursor()
    try:
        c.executemany('''
        INSERT INTO employees (id,name,age,dept)
        VALUES (?,?,?,?)
        ''', (employees,) )
        conn.commit()
    except sqlite3.Error as e:
        conn.rollback()
        print(e)
    finally:
        conn.close()


employees = [
    (6, 'Frank', 40, 'Finance'),
    (7, 'Grace', 29, 'Engineering'),
    (8, 'Hannah', 35, 'Marketing'),
    (9, 'Ivan', 38, 'Sales'),
    (6, 'Jack', 45, 'HR')  # Duplicate ID to cause an error
]

insert_many_val(employees)



```

```python
#assignment 7 2
def insert_many_val(updates):
    conn = sqlite3.connect('test.db')
    c=conn.cursor()
    try:
        c.executemany('''
        UPDATE employees
        SET age=?
        WHERE id =? 
        ''', (updates,) )
        conn.commit()
    except sqlite3.Error as e:
        conn.rollback()
        print(e)
    finally:
        conn.close()

updates = [
    (32, 1),
    (26, 2),
    (33, 3),
    (41, 4),  # Non-existing ID to cause an error
    (23, 5)
]
insert_many_val(updates)
```
