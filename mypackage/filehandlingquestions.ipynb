```python
#assignment 1 
def read (file_name):
    with open (file_name,'r') as file:
        for lines in file:
            print ( lines.strip())

read('sample.txt')
```

```python
#assignment 2
lst=['str1\n','str2\n','str3\n']
def write_str(file_name):
    with open (file_name,'w') as file :
        file.writelines(lst)

write_str('output.txt')
```

```python
#assignment 3
import os 
def copy_contents (source_file, destination_file):
    with open (source_file,'r')as src:
        with open (destination_file,'w')as dest:
            dest.write(src.read())

copy_contents('source.txt','destination.txt')

print("Current folder:", os.getcwd())
print("Files here:", os.listdir())

with open('source.txt', 'r') as src:
    content = src.read()
    print(content)
```

```python
#assignment 4 
def append ( file_name , str ):
    with open ( file_name , 'a')as file :
        file.write (str + "\n")

append ('log.txt','amaan is good')
```

```python
#assignment 5 
def count (file_name ):
    with open (file_name, 'r')as file:
        content=file.read()
        words = content.split()
        return len(words)

print(count("document.txt"))
```

```python
#assignment 6
def replace (file_name , word ,new  ):
    with open (file_name , 'r') as file :
        content = file.read ()
    text=content.replace (word,new)
    with open (file_name ,'w') as file :
        file.write (text)

replace ("data.txt","amaan", "josh")
```

```python
#assignment 7

def reverse (file_name ):
    with open (file_name,'r') as file :
         lines = file.readlines()
    for line in reversed(lines ):
         print (line.strip())

reverse("reverse.txt")
```

```python
#assignment 8
def count (file_name):
    with open (file_name, 'r') as file :
        lines = file.readlines()
        words = sum(len(line.split()) for line in lines)
        char = sum (len(line)for line in lines)
    return len(lines), words , char 

print (count ("stats.txt"))
```

```python
#assignment 9

def merge_files(lst_of_files , merged_file):
    with open (merged_file,'w') as outfile :
        for fname in lst_of_files:
            with open (fname , 'r')as infile :
                outfile.write(infile.read() + '\n')

merge_files(['log.txt','output.txt'], 'merged.txt')
```
